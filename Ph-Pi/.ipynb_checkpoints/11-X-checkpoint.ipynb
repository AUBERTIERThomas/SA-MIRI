```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_epoch_time(bs, time, w=0.7):
    fake_x_axis = [i for i in range(len(bs))]
    fig, ax = plt.subplots()
    ax.bar(fake_x_axis, time, width=w)
    ax.set_title('Epoch Time')
    ax.set_xlabel('Batch size (img)')
    ax.set_ylabel('Time (s)')
    ax.set_xticks(fake_x_axis)
    ax.set_xticklabels(bs)
    r = (max(time)-min(time))/5
    ax.set_ylim([min(time)-r, max(time)+r])
    plt.grid(True)
    plt.show()

def plot_throughput(bs, thr, w=0.7):
    fake_x_axis = [i for i in range(len(bs))]
    fig, ax = plt.subplots()
    ax.bar(fake_x_axis, thr, width=w)
    ax.set_title('Throughput per second')
    ax.set_xlabel('Batch size (img)')
    ax.set_ylabel('Throughput/s')
    ax.set_xticks(fake_x_axis)
    ax.set_xticklabels(bs)
    r = (max(thr)-min(thr))/5
    ax.set_ylim([min(thr)-r, max(thr)+r])
    plt.grid(True)
    plt.show()

def plot_nm_workers(nw, thr, w=0.7):
    fake_x_axis = [i for i in range(len(nw))]
    fig, ax = plt.subplots()
    ax.bar(fake_x_axis, thr, width=w)
    ax.set_title('Throughput per second')
    ax.set_xlabel('Number of workers')
    ax.set_ylabel('Throughput/s')
    ax.set_xticks(fake_x_axis)
    ax.set_xticklabels(nw)
    r = (max(thr)-min(thr))/5
    ax.set_ylim([min(thr)-r, max(thr)+r])
    plt.grid(True)
    plt.show()

def plot_final(id, time, title, w=0.7):
    fake_x_axis = [i for i in range(len(id))]
    fig, ax = plt.subplots(figsize=(12,6))
    ax.bar(fake_x_axis, time, width=w)
    ax.set_title(title)
    ax.set_xlabel('Conditions')
    ax.set_ylabel('Time (s)')
    ax.set_xticks(fake_x_axis)
    ax.set_xticklabels(id)
    r = (max(time)-min(time))/5
    ax.set_ylim([min(time)-r, max(time)+r])
    plt.grid(True)
    plt.show()
```

```python
bs = [16,32,64]
time = [906,758,732]
thr = [79,85,88]

plot_epoch_time(bs, time)
```

```python
plot_throughput(bs, thr)
```

```python
bs = [32,64,128,256,512]
time = [1085,297,272,263,262]
thr = [1443,1732,1868,1919,1922]

plot_epoch_time(bs, time)
```

```python
plot_throughput(bs, thr)
```
