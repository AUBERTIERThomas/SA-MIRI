```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_epoch_time(GPU, time):
    plt.bar(GPU, time)
    plt.plot(GPU, time, color="red")
    plt.title('Epoch Time')
    plt.xlabel('GPUs')
    plt.ylabel('Time (s)')
    plt.show()

def plot_throughput(GPU, time, nb_im=50000):
    plt.bar(GPU, nb_im/np.array(time))
    plt.plot(GPU, nb_im/np.array(time), color="red")
    plt.title('Throughput per second (50000 images)')
    plt.xlabel('GPUs')
    plt.ylabel('Throughput/s')
    plt.show()

def plot_speedup(GPU, time):
    plt.bar(GPU, GPU, color="green", label="Linear speedup")
    plt.bar(GPU, time[0]/np.array(time), color="red", label="Measured speedup")
    plt.title('Speedup')
    plt.xlabel('GPUs')
    plt.ylabel('Speedup')
    plt.legend()
    plt.show()

def plot_efficiency(GPU, time):
    plt.bar(GPU, (time[0]/np.array(time))/GPU * 100)
    plt.title('Efficiency')
    plt.xlabel('GPUs')
    plt.ylabel('Efficiency')
    plt.show()
```

### 1/2/4 GPUs

```python
GPU = [1,2,4]
time = [13,8,5]

plot_epoch_time(GPU, time)
```

```python
plot_throughput(GPU, time, nb_im=50000)
```

```python
plot_speedup(GPU, time)
```

```python
plot_efficiency(GPU, time)
```

```python
def plot_epoch_time_2(GPU, time_50, time_152, time_102=None):
    plt.plot(GPU, time_50, color="red", label="ResNet50V2")
    plt.plot(GPU, time_152, color="blue", label="ResNet152V2")
    if time_102:
        plt.plot(GPU, time_102, color="green", label="ResNet102V2")
    plt.title('Epoch Time')
    plt.xlabel('GPUs')
    plt.ylabel('Time (s)')
    plt.xticks(GPU)
    plt.legend()
    plt.show()

def plot_throughput_2(GPU, time_50, time_152, time_102=None, nb_im=50000):
    plt.plot(GPU, nb_im/np.array(time_50), color="red", label="ResNet50V2")
    plt.plot(GPU, nb_im/np.array(time_152), color="blue", label="ResNet152V2")
    if time_102:
        plt.plot(GPU, nb_im/np.array(time_102), color="green", label="ResNet102V2")
    plt.title('Throughput per second (50000 images)')
    plt.xlabel('GPUs')
    plt.ylabel('Throughput/s')
    plt.xticks(GPU)
    plt.legend()
    plt.show()

def plot_speedup_2(GPU, time_50, time_152, time_102=None):
    plt.plot(GPU, time_50[0]/np.array(time_50), color="red", label="ResNet50V2")
    plt.plot(GPU, time_152[0]/np.array(time_152), color="blue", label="ResNet152V2")
    if time_102:
        plt.plot(GPU, time_102[0]/np.array(time_102), color="green", label="ResNet102V2")
    plt.plot(GPU, GPU, "--", color="black", label="Linear speedup")
    plt.title('Speedup')
    plt.xlabel('GPUs')
    plt.ylabel('Speedup')
    plt.xticks(GPU)
    plt.legend()
    plt.show()

def plot_efficiency_2(GPU, time_50, time_152, time_102=None):
    plt.plot(GPU, time_50[0]/np.array(time_50)/GPU * 100, color="red", label="ResNet50V2")
    plt.plot(GPU, time_152[0]/np.array(time_152)/GPU * 100, color="blue", label="ResNet152V2")
    if time_102:
        plt.plot(GPU, time_102[0]/np.array(time_102)/GPU * 100, color="green", label="ResNet102V2")
    plt.title('Efficiency')
    plt.xlabel('GPUs')
    plt.ylabel('Efficiency')
    plt.xticks(GPU)
    plt.legend()
    plt.show()
```

```python
GPU = [1,2,4]
time_50 = [13,8,5]
time_152 = [32,18,10]

plot_epoch_time_2(GPU, time_50, time_152)
```

```python
plot_throughput_2(GPU, time_50, time_152)
```

```python
plot_speedup_2(GPU, time_50, time_152)
```

```python
plot_efficiency_2(GPU, time_50, time_152)
```
